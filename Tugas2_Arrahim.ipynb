##Task 8
Sekarang kita mulai training nya

```python
# ==========================================
# Task 8
# Sekarang kita mulai training nya
# ==========================================

from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, GradientBoostingClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# Split data training menjadi train & validation
x_train, x_val, y_train, y_val = train_test_split(
    xtrain, ytrain, test_size = .2, random_state = 21, stratify = ytrain
)

x_train = x_train.reset_index(drop = True)
x_val = x_val.reset_index(drop = True)
y_train = y_train.reset_index(drop = True)
y_val = y_val.reset_index(drop = True)

# 1. Random Forest
rf = RandomForestClassifier(n_jobs = -1, random_state = 0, n_estimators = 1500, max_depth = 7)
rf.fit(x_train, y_train)
yhat_rf = rf.predict(x_val)
score_rf = accuracy_score(y_val, yhat_rf)
print(f'RF Score : {score_rf}')

# 2. XGBoost
xgb = XGBClassifier(
    n_estimators = 2000, n_jobs = -1, random_state = 0, max_depth = 7,
    learning_rate = 0.1, reg_alpha = 0, reg_lambda = 0,
    colsample_bytree = 1, colsample_bynode = 1, colsample_bylevel = 0.1,
    subsample = 0.7, tree_method = 'exact', objective = 'multiclass'
)
xgb.fit(x_train, y_train)
yhat_xgb = xgb.predict(x_val)
score_xgb = accuracy_score(y_val, yhat_xgb)
print(f'XGBoost Score : {score_xgb}')

# 3. Gradient Boosting
gb = GradientBoostingClassifier(n_estimators = 1000, random_state = 0, max_depth = 7)
gb.fit(x_train, y_train)
yhat_gb = gb.predict(x_val)
score_gb = accuracy_score(y_val, yhat_gb)
print(f'GBoost Score : {score_gb}')

# 4. LightGBM
lgbm = LGBMClassifier(
    n_estimators = 2000, n_jobs = -1, random_state = 0, max_depth = 7,
    learning_rate = 0.1, reg_alpha = 0, reg_lambda = 0,
    colsample_bytree = 1, colsample_bynode = 1, colsample_bylevel = 0.1,
    subsample = 0.7, objective = 'multiclass', verbose = -1
)
lgbm.fit(x_train, y_train)
yhat_lgbm = lgbm.predict(x_val)
score_lgbm = accuracy_score(y_val, yhat_lgbm)
print(f'LGBM Score : {score_lgbm}')
```

```python
# 2) Lakukan evaluasi model dengan K-fold
# Saya akan melakukan training di K-Fold dan menyimpan hasilnya di sini

from sklearn.model_selection import RepeatedStratifiedKFold
from tqdm.auto import tqdm
import numpy as np

# Menyiapkan K-Fold (5 folds x 3 pengulangan = 15 iterasi)
kf = RepeatedStratifiedKFold(n_splits = 5, random_state = 21, n_repeats = 3)

# Menyiapkan tempat penyimpanan instance model dan skor
models_rf, models_xgb, models_gb, models_lgbm = [], [], [], []
scores_xgb, scores_rf, scores_gb, scores_lgbm = [], [], [], []

for i, (t, v) in enumerate(tqdm(kf.split(xtrain, ytrain), total = 15)):
    xt = xtrain.iloc[t, :]
    xv = xtrain.iloc[v, :]
    yt = ytrain.iloc[t]
    yv = ytrain.iloc[v]

    # Fit/Latih semua model
    rf.fit(xt, yt)
    xgb.fit(xt, yt)
    gb.fit(xt, yt)
    lgbm.fit(xt, yt)

    # Prediksi ke data validasi fold
    yhat_xgb = xgb.predict(xv)
    yhat_rf = rf.predict(xv)
    yhat_gb = gb.predict(xv)
    yhat_lgbm = lgbm.predict(xv)

    # Simpan Skor Akurasi
    scores_xgb.append(accuracy_score(yv, yhat_xgb))
    scores_rf.append(accuracy_score(yv, yhat_rf))
    scores_gb.append(accuracy_score(yv, yhat_gb))
    scores_lgbm.append(accuracy_score(yv, yhat_lgbm))

    # Simpan Instance Model
    models_rf.append(rf)
    models_xgb.append(xgb)
    models_gb.append(gb)
    models_lgbm.append(lgbm)

# Menampilkan Rata-rata Skor Akurasi K-Fold
print(f'Mean RF Score   : {np.mean(scores_rf)}')
print(f'Mean XGB Score  : {np.mean(scores_xgb)}')
print(f'Mean GB Score   : {np.mean(scores_gb)}')
print(f'Mean LGBM Score : {np.mean(scores_lgbm)}')
```

*Membagi data menjadi set Training (80%) dan Testing (20%) secara stratified. Melatih 4 algoritma tree (Random Forest, XGBoost, Gradient Boosting, dan LightGBM) menggunakan RepeatedStratifiedKFold (15 iterasi) untuk mendapatkan perkiraan performa yang stabil.*

#Task 9
Training sudah selesai dan model sudah diketahui, kita uji hasilnya dengan data testing untuk melihat kemungkinan overfitting dan underfitting, sekaligus menghitung kualitas model dengan  pengukuran.


```python
# ==========================================
# Task 9
# Model Blending (Ensemble) & Evaluasi Akhir pada Data Testing
# ==========================================

import scipy
import numpy as np

# 1. Mengumpulkan prediksi dari seluruh model terlayani di K-Fold untuk data xtest
preds_rf = np.array([m.predict(xtest) for m in models_rf])
preds_xgb = np.array([m.predict(xtest) for m in models_xgb])
preds_gb = np.array([m.predict(xtest) for m in models_gb])
preds_lgbm = np.array([m.predict(xtest) for m in models_lgbm])

# 2. Penggabungan prediksi (Majority Voting / Blending)
all_preds = np.concatenate([preds_rf, preds_xgb, preds_gb, preds_lgbm], axis=0)
final_preds, _ = scipy.stats.mode(all_preds, axis=0)
final_preds = final_preds.ravel()

# 3. Evaluasi Skor Akurasi Akhir
final_acc = accuracy_score(ytest, final_preds)
print(f"Akurasi Final Model Blending (pada xtest): {final_acc:.4f}")

# 4. Visualisasi Confusion Matrix Akhir
cm = confusion_matrix(ytest, final_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues')
```

```python
from sklearn.metrics import classification_report

# Menampilkan Laporan Evaluasi Lengkap
print("=== Classification Report Final Model Blending ===")
print(classification_report(ytest, final_preds, target_names=['Gagal', 'Lulus Tidak Tepat Waktu', 'Lulus Tepat Waktu']))
```
